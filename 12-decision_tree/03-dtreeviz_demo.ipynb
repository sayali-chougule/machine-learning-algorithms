```python
import graphviz.backend as be

from sklearn.datasets import *
from dtreeviz.trees import *
from IPython.display import Image, display_svg, SVG
from sklearn import tree
```

```python
clas = tree.DecisionTreeClassifier()
iris = load_iris()

import pandas as pd

X_train = pd.DataFrame(iris.data, columns=iris.feature_names)
y_train = iris.target
clas.fit(X_train, y_train)
```

```python
from sklearn.tree import plot_tree
```

```python
plot_tree(clas)
```

### Classification

```python
from dtreeviz import dtreeviz
```

```python
import dtreeviz
viz = dtreeviz.model(clas,
               X_train,
               y_train,
               target_name='species',
               feature_names = iris.feature_names,
               class_names = ['setosa', 'versicolor', 'virginica'],
            #    scale = 1.5
               )

viz.view()
```

### Regression

```python
from sklearn.datasets import fetch_california_housing
```

```python
regr = tree.DecisionTreeRegressor(max_depth=2)
cal = fetch_california_housing()
```

```python
X_train = cal.data
y_train = cal.target   
regr.fit(X_train, y_train)

viz = dtreeviz.model(regr,
               X_train,
               y_train,
               target_name='Species',
               feature_names= cal.feature_names
               )

viz.view()
```

### Horizontal Decision Tree

```python
X_train = pd.DataFrame(iris.data, columns=iris.feature_names)
y_train = iris.target

clas.fit(X_train, y_train)

viz = dtreeviz.model(
    clas,
    X_train,
    y_train,
    target_name='species',
    feature_names=iris.feature_names,
    class_names=list(iris.target_names)
)

viz.view(orientation="LR")
```

### Show Prediction Path

```python
clas = tree.DecisionTreeClassifier()
iris = load_iris()

X_train = iris.data
y_train = iris.target
clas.fit(X_train, y_train)

X = iris.data[np.random.randint(0, len(iris.data)), :]

viz = dtreeviz.model(clas,
                     X_train,
                     y_train,
                     feature_names=iris.feature_names,
                     class_names=['setosa', 'versicolor', 'virgenica'],
                    #  x=X
                    )

viz.view(x=X)
```
