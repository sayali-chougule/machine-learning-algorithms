#### Problem 1 

Outlook = Sunny, Temp = Hot, Humidity = High, Wind = Weak

Play or No Play ?

#### Solution

P(Yes | Sunny, Hot, High, Weak) = P(Sunny | Yes) * P(Hot | Yes) * P(High | Yes) * P(Weak | Yes)

P(No | Sunny, Hot, High, Weak) = P(Sunny | No) * P(Hot | No) * P(High | No) * P(Weak | No)

Compare using Maximum Posteriori Rule

```python
P_Yes = py * P_Sunny_y * P_Hot_y * P_High_y * P_Weak_y 
print(P_Yes)
```

```python
P_No = pn * P_Sunny_n * P_Hot_n * P_High_n * P_Weak_n
print(P_No)
```

### Result

No Play 

As P_No > P_Yes

#### Problem 2

#### Solution

P(Yes | Overcast, Cold, Low, Weak) = P(Overcast | Yes) * P(Cold | Yes) * P(Low | Yes) * P(Weak | Yes)

P(No | Overcast, Cold, Low, Weak) = P(Overcast | No) * P(Cold | No) * P(Low | No) * P(Weak | No)

Compare using Maximum Posteriori Rule

```python
P_Yes = py * P_overcast_y * P_Cool_y * P_Normal_y * P_Weak_y 
print(P_Yes)
```

```python
P_No = pn * P_overcast_n * P_Cool_n * P_Normal_n * P_Weak_n
print(P_No)
```

### Result

As P_Yes > P_No
