```python
import numpy as np
import pandas as pd
```

```python
data = {'R&D Spend': [165.0, 101.0, 29.0], 'Administration': [137.0, 92.0, 127.0], 'Marketing Spend':[472.0, 250.0, 201.0], 'Profit':[192.0, 144.0, 91.0]}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df['f0(X)'] = df['Profit'].mean()
```

```python
df
```

```python
df['ri1'] = df['Profit'] - df['f0(X)']
```

```python
df
```

```python
from sklearn.tree import DecisionTreeRegressor
dt1 = DecisionTreeRegressor(max_depth=1)
```

```python
dt1.fit(df.iloc[:, 0:3].values, df['ri1'].values)
```

```python
from sklearn.tree import plot_tree
plot_tree(dt1)
```
